```python
import pandas as pd
import pickle

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import FunctionTransformer

# Load your movie data
movies_dict = pickle.load(open('model/movies_dict.pkl', 'rb'))
movies = pd.DataFrame(movies_dict)

# Handle missing tags
movies['tags'] = movies['tags'].fillna('')

# Create a text-processing pipeline
pipeline = Pipeline([
    ('vectorizer', CountVectorizer(max_features=5000, stop_words='english'))
])

# Transform movie tags into vectors
vectors = pipeline.fit_transform(movies['tags'])

# Calculate similarity between movies
similarity = cosine_similarity(vectors)

print(vectors.shape)
print(similarity.shape)
```

```python
def recommend(movie_title, movies, similarity, top_n=5):
    matches = movies.index[
        movies['title'].str.casefold() == movie_title.casefold()
    ]

    if len(matches) == 0:
        return []

    index = matches[0]

    distances = sorted(
        enumerate(similarity[index]),
        key=lambda x: x[1],
        reverse=True
    )

    recommendations = []

    for movie_index, score in distances:
        if movie_index == index:
            continue

        recommendations.append({
            'title': movies.iloc[movie_index]['title'],
            'similarity_score': float(score)
        })

        if len(recommendations) == top_n:
            break

    return recommendations
```

```python
recommend('The Dark Knight', movies, similarity)
```
